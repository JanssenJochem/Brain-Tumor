Step 1 — Inspect the dataset and count the images
This gives you the basic dataset inventory and reveals unexpected file types.

```python
from pathlib import Path
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

# Change this to the directory containing your four class folders
DATA_DIR = Path("../archive/data")

classes = ["normal", "glioma_tumor", "meningioma_tumor", "pituitary_tumor"]
valid_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

records = []
unexpected_files = []

for class_name in classes:
    class_dir = DATA_DIR / class_name

    if not class_dir.is_dir():
        print(f"WARNING: Folder not found: {class_dir}")
        continue

    for file_path in class_dir.rglob("*"):
        if not file_path.is_file():
            continue

        if file_path.suffix.lower() not in valid_extensions:
            unexpected_files.append(str(file_path))
            continue

        records.append({
            "filename": file_path.name,
            "filepath": str(file_path),
            "class": class_name,
            "extension": file_path.suffix.lower()
        })

df = pd.DataFrame(records)

# --- Text Summaries ---
print("Total number of images:", len(df))
print("\nImages per class:")
print(df["class"].value_counts().reindex(classes, fill_value=0))

print("\nFile formats:")
print(df["extension"].value_counts())

print("\nUnexpected files:", len(unexpected_files))
if len(unexpected_files) > 0:
    print(unexpected_files[:20])

# --- Visual 1: Class Distribution Bar Plot ---
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x="class", order=classes, palette="Blues_r")
plt.title("Image Count per Class (Checking Imbalance)")
plt.xlabel("Class Name")
plt.ylabel("Number of Images")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# --- Visual 2: Sample Image Grid (One per class) ---
fig, axes = plt.subplots(1, len(classes), figsize=(16, 5))

for i, class_name in enumerate(classes):
    class_subset = df[df["class"] == class_name]
    
    if len(class_subset) > 0:
        # Pick the first image path for this class
        sample_path = class_subset.iloc[0]["filepath"]
        img = Image.open(sample_path)
        
        axes[i].imshow(img, cmap="gray" if img.mode == "L" else None)
        axes[i].set_title(f"{class_name}\n(Size: {img.size})")
        axes[i].axis("off")
    else:
        axes[i].set_title(f"{class_name}\n(No images found)")
        axes[i].axis("off")

plt.tight_layout()
plt.show()
```
